```python
top_products = df.groupby("Product Name")["Sales"].sum()

top_products = top_products.sort_values(
    ascending=False
).head(10)

top_products
```

```python
top_products.plot(
    kind="bar",
    figsize=(10,5)
)

plt.title("Top 10 Selling Products")
plt.xlabel("Product")
plt.ylabel("Sales")

plt.show()
```

```python
profit_by_category = df.groupby(
    "Category"
)["Profit"].sum()

profit_by_category
```

```python
profit_by_category.plot(
    kind="bar",
    figsize=(8,5)
)

plt.title("Profit by Category")
plt.xlabel("Category")
plt.ylabel("Profit")

plt.show()
```

```python
region_profit = df.groupby(
    "Region"
)["Profit"].sum()

region_profit
```

```python
region_profit.plot(
    kind="bar",
    figsize=(8,5)
)

plt.title("Profit by Region")
plt.xlabel("Region")
plt.ylabel("Profit")

plt.show()
```

```python
category_sales = df.groupby(
    "Category"
)["Sales"].sum()

category_sales
```

```python
category_sales.plot(
    kind="pie",
    autopct="%1.1f%%",
    figsize=(7,7)
)

plt.title("Sales Contribution by Category")

plt.ylabel("")

plt.show()
```
